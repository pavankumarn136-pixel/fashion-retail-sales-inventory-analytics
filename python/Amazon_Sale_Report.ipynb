```python
matched_sales = sales_inventory[
    sales_inventory['inventory_match'] == 'both'
].copy()

matched_sales['stock_status'].value_counts()
```

```python
demand_sales = matched_sales[
    matched_sales['status_group'].isin(
        ['Delivered', 'Shipped / In Process']
    )
].copy()

demand_sales.shape
```

```python
restock_priority = (
    demand_sales
    .groupby(
        [
            'sku',
            'category',
            'design_no',
            'color',
            'stock',
            'stock_status'
        ],
        as_index=False
    )
    .agg(
        units_sold=('qty', 'sum'),
        sales_revenue=('amount', 'sum'),
        sales_records=('order_id', 'count')
    )
)

restock_priority = (
    restock_priority[
        restock_priority['stock_status'].isin(
            ['Low Stock', 'Out of Stock']
        )
    ]
    .sort_values(
        ['stock', 'units_sold'],
        ascending=[True, False]
    ))
```

```python
restock_priority.head(10)
```

```python
category_restock_risk = (
    restock_priority
    .groupby('category', as_index=False)
    .agg(
        at_risk_skus=('sku', 'nunique'),
        demand_units=('units_sold', 'sum'),
        historical_sales_revenue=('sales_revenue', 'sum')).sort_values(  'historical_sales_revenue',
        ascending=False))
```

```python
category_restock_risk
```

```python
out_of_stock_category_risk = (restock_priority[restock_priority['stock_status'] == 'Out of Stock'].groupby('category', as_index = False).agg(
    out_of_stock_skus = ('sku','nunique'),
    demand_units = ('units_sold','sum'),
    historical_sales_revenue = ('sales_revenue','sum')).sort_values('historical_sales_revenue',ascending = False))
```

```python
delivered_kpis = pd.Series({
    'delivered_orders': delivered_revenue_sales['order_id'].nunique(),
    'units_sold': delivered_revenue_sales['qty'].sum(),
    'delivered_revenue': delivered_revenue_sales['amount'].sum()
})
```

```python
delivered_kpis
```

```python
delivered_order_summary = ( delivered_revenue_sales.groupby('order_id', as_index = False).agg(
    order_revenue = ('amount', 'sum'),
    order_units = ('qty','sum')))

average_order_value = delivered_order_summary['order_revenue'].mean()
```

```python
average_order_value
```

```python
monthly_delivered_sales = (
    delivered_revenue_sales.groupby(['month_number', 'month_name'], as_index = False).agg(
        delivered_orders = ('order_id', 'nunique'),
        units_sold = ('qty','sum'),
        revenue = ('amount', 'sum')).sort_values('month_number')
)
```

```python
monthly_delivered_sales
```

```python
monthly_date_coverage = (
    delivered_revenue_sales
    .groupby(['month_number', 'month_name'], as_index=False)
    .agg(
        first_sale_date=('date', 'min'),
        last_sale_date=('date', 'max'),
        days_with_sales=('date', 'nunique')
    )
    .sort_values('month_number')
)

monthly_date_coverage
```

```python
monthly_sales_daily = monthly_delivered_sales.merge(monthly_date_coverage[['month_number', 'days_with_sales']],
                                                    on = 'month_number',
                                                    how = 'left')

monthly_sales_daily['revenue_per_day'] = (monthly_sales_daily['revenue'] / 
                                          monthly_sales_daily['days_with_sales']).round(2)
```

```python
import pandas as pd
from pathlib import Path

project_folder = Path(
    r'C:\Users\pavan\OneDrive\Desktop\E-Commerce SalesDataset'
)

amazon_clean = pd.read_csv(
    project_folder / 'Amazon_Sales_Cleaned.csv',
    parse_dates=['date'],
    low_memory=False
)

inventory_clean = pd.read_csv(
    project_folder / 'Inventory_Cleaned.csv',
    low_memory=False
)

amazon_clean['sku'] = amazon_clean['sku'].astype('string').str.strip()
inventory_clean['sku'] = inventory_clean['sku'].astype('string').str.strip()

inventory_merge_ready = inventory_clean[
    inventory_clean['sku_quality'] == 'Valid SKU'
].copy()

inventory_for_merge = inventory_merge_ready[
    ['sku', 'design_no', 'stock', 'color', 'stock_status', 'sku_quality']
].copy()

sales_inventory = amazon_clean.merge(
    inventory_for_merge,
    on='sku',
    how='left',
    validate='many_to_one',
    indicator='inventory_match'
)

matched_sales = sales_inventory[
    sales_inventory['inventory_match'] == 'both'
].copy()

demand_sales = matched_sales[
    matched_sales['status_group'].isin(
        ['Delivered', 'Shipped / In Process']
    )
].copy()

delivered_sales = sales_inventory[
    sales_inventory['status_group'] == 'Delivered'
].copy()

delivered_revenue_sales = delivered_sales.dropna(
    subset=['amount']
).copy()
```
